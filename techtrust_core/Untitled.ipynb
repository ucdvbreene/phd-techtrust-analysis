```python
import pandas as pd
data = {
    'Participant Number': [1, 1, 2, 2],
    'Topic Quote': [
        "I like using this technology",
        "It's okay, could be better",
        "Very helpful tool",
        "Not what I expected"
    ],
    'Adaptability': [1, 0, 1, 0],
    'Intent to use': [0, 1, 1, 0],
    'System Performance': [1, 1, 0, 0],
    'System Understanding': [0, 1, 1, 1],
    'Openness to Experimentation': [1, 0, 0, 1],
    'Percieved Usefulness': [1, 0, 1, 0]
}

df = pd.DataFrame(data)
```

```python
df_with_sentiment = compute_sentiment(df)
df_with_sentiment
```

```python
factor_cols = ['Adaptability', 'Intent to use', 'System Performance',
               'System Understanding', 'Openness to Experimentation',
               'Percieved Usefulness']

weights = [0.25, 0.15, 0.2, 0.1, 0.15, 0.15]  # must sum to 1

df_with_techtrust = compute_technology_trust(df_with_sentiment, factor_cols, weights)
df_with_techtrust[['Participant Number', 'TechTrust']]
```
